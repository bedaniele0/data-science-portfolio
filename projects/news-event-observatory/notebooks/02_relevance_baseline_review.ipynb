# 02 Relevance Baseline Review

Inspect the rule-based relevance filter from Component 2. The baseline is intentionally explainable and should be improved with human labels later.


```python
from pathlib import Path
import json
import pandas as pd

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent

pd.set_option("display.max_colwidth", 140)
ROOT
```

```python
scored = pd.read_csv(ROOT / "data/processed/scored_articles.csv", dtype=str, keep_default_na=False)
report = json.loads((ROOT / "reports/relevance_report.json").read_text())
report
```

```python
scored["is_relevant_bool"] = scored["is_relevant"].astype(str).str.lower().eq("true")
scored["is_relevant_bool"].value_counts().rename_axis("is_relevant").reset_index(name="articles")
```

```python
cols = ["article_id", "query", "title", "media", "published_at", "relevance_score", "matched_event_terms", "matched_actor_terms", "url"]
scored.loc[scored["is_relevant_bool"], cols]
```

```python
# Useful review sample: event terms without actor terms. These are likely candidates for recall/precision tuning.
scored.loc[
    scored["has_event_terms"].astype(str).str.lower().eq("true")
    & ~scored["has_actor_terms"].astype(str).str.lower().eq("true"),
    ["article_id", "query", "title", "media", "published_at", "matched_event_terms", "url"],
].head(25)
```
