# 00 Project Overview

Portfolio-facing overview of Media Event Intelligence. This notebook summarizes the current pipeline state and the main deliverables generated by the reproducible code in `src/`.


```python
from pathlib import Path
import json
import pandas as pd

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent

pd.set_option("display.max_colwidth", 140)
ROOT
```

```python
reports = {}
for name in ["ingestion_report", "relevance_report", "clustering_report"]:
    path = ROOT / "reports" / f"{name}.json"
    reports[name] = json.loads(path.read_text()) if path.exists() else {"missing": str(path)}
reports
```

```python
summary = pd.DataFrame([
    {
        "component": "1_ingestion",
        "main_output": "data/interim/normalized_articles.csv",
        "rows": reports["ingestion_report"].get("normalized_rows"),
        "status": "done" if "normalized_rows" in reports["ingestion_report"] else "missing",
    },
    {
        "component": "2_relevance",
        "main_output": "data/processed/relevant_articles.csv",
        "rows": reports["relevance_report"].get("relevant_articles"),
        "status": "done" if "relevant_articles" in reports["relevance_report"] else "missing",
    },
    {
        "component": "3_clustering",
        "main_output": "data/processed/events.csv",
        "rows": reports["clustering_report"].get("total_events"),
        "status": "done" if "total_events" in reports["clustering_report"] else "missing",
    },
])
summary
```

```python
events_path = ROOT / "data/processed/events.csv"
events = pd.read_csv(events_path) if events_path.exists() else pd.DataFrame()
events[["event_id", "event_title", "article_count", "media_count", "confidence", "status"]].head(10)
```
